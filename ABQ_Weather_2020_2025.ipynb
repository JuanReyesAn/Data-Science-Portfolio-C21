```
folder_path = '/content/drive/MyDrive/ddds-cohort-21/Projects/Capstone/ABQ Sunport Weather'

#Use glob to find all files ending in .csv inside folder
csv_files = glob.glob(os.path.join(folder_path, "*.csv"))

#Read each CSV into a list of DataFrames
df_list = []
for file in csv_files:
    # Read the current CSV file
    df = pd.read_csv(file)
    df_list.append(df)

#Combine all individual DataFrames into one large DataFrame
weather_df = pd.concat(df_list, ignore_index=True)
weather_df.head()
```

## Cleaning

```
# assign a variable to the data from sunport so we can solely focus on that portion and begin cleaning that
sunport_w = weather_df[weather_df['STATION'] == 'USW00023050']

# Reset index so it starts cleanly from 0
sunport_w.reset_index(drop=True, inplace=True)
```

```
# this data has a lot of nulls will beging by fuguring out what columns are 100% nulls if any then drop.
sunport_w = sunport_w.dropna(axis=1, how='all')
```
